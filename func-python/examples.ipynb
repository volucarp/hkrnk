Create an Iterator via a class

````python
class Alphabet():
    
    chars = 'abcde'
     
    def __init__(self): 
        # Number of characters read 
        self.pos = 0 

    def __iter__(self): 
        return self 

    def __next__(self): 
        if self.pos < len(self.chars): 
            c = self.chars[self.pos] 
            self.pos += 1 
            return c 
        else: 
            raise StopIteration  

for c in Alphabet(): 
    print(c)
````

````python
#prefer classes for complex initializations 

class Format(): 

  def __init__(self): 
    self.p = 0 
    self.w = 0 

  def prec(self, n): 
    self.p = n 
    return self 

  def width(self, n): 
    self.w = n 
    return self 

  def __call__(self, x): 
    return '{:{width}.{prec}f}'.format(x, width=self.w, prec=self.p)

format3 = Format().width(10).prec(3) 
print(format3(1.2345678))  
````

An advantage of generator comprehensions is that they can directly replace standard functions like map. This is useful if you want a slight variant. Here is a comprehension equivalent of a simple one-parameter map of function, fn, over iterable, it:

````python
def fn(x):
    return x+1;

it = [1,2,3,4]

a = map(fn, it)
b = (fn(x) for x in it)

print(list(a), list(b)) # to view the result
````

````python
#map and variant together
def fn(x):
    return x+1;

it = [1,2,3,4]

a = map(lambda x: fn(x)+1, it)
b = (fn(x)+1 for x in it)

print(list(a), list(b)) # to view the result
````

## Partial and Curring

Partial application is a way of creating a new function based on an existing function but with some of the parameters already filled in with a chosen value. For example, here is a closure that creates partial applications of the standard function, `max`:

````python
def maxn(n):
  def f(x):
      return max(n, x)
  return f

max0 = maxn(0)
print(max0(3)) # Equivalent to max(0, 3) -> 3 
print(max0(-1)) # Equivalent to max(0, -1) -> 0
````

````python
#quadratic function

def quad(a, b, c, x):
  return a*x*x + b*x + c

def quad_abc(a, b, c):
    def f(x):
        return quad(a, b, c, x)
    return f

myquad = quad_abc(1, -3, 2)

print(myquad(0))  # 2
print(myquad(1))  # 0
print(myquad(2))  # 0
````

````python
from functools import partial

max0 = partial(max, 0)

print(max0(3)) # 3 
print(max0(-1)) # 0
````

## Caveat: Empty list

One final point is that the reduce function will give an error if it is called with an empty list. If we want to avoid this, we need to supply a suitable initial value as a third parameter. But what should we use?


The initial value should be the identity value for the operation we are using. If we were using the add operation, we would use 0 because x plus 0 is x. If we were using the mul operation, it would be 1 because x times 1 is x. In the case of composition, we want a value that acts as an identity value when composed with any function. This value is the function, f(x), that returns x. Hence, to avoid the error, change the last line of compose to:

````python
from functools import partial, reduce
from operator import add, mul

def compose(*f):
  def compose2(f, g):
    def fn(x):
      return f(g(x))
    return fn
  return reduce(compose2, f, lambda x: x) ## changed here

my_composite_function = compose (partial(add, 1), partial(add, 2), partial(mul, 3))
print(my_composite_function(1))
````

We will concentrate on PyMonad in this section. It provides composition of curried functions (see the PyMonad @curry decorator described earlier in this chapter).

Rather than using a compose function, PyMonad uses a compose operator, *.

Here is how it works.

````python
from pymonad import curry

@curry
def reversedc(x):
   return reversed(x)

@curry
def rangec(n):
   return range(n)

countdown = reversedc * rangec 

my_iterator = countdown(9)
print(my_iterator) # a range_iterator object 
print(next(my_iterator)) # prints 8
````

##What is a monad?

A monad wraps a value in a similar way to a functor. However, a monad has an additional function, called bind, that:

    Accepts a single parameter
    Returns a value wrapped in a monad

Unlike map, the bind function, itself, is responsible for wrapping the return value. This means that the bind function can decide what sort of monad to wrap the result in.

Here is an example, the function, oneover, returns $\frac{1}{x}​$​​, wrapped in a Just monad

````python
from oslash import Just, Nothing

def oneover(x):
    try:
        ret = 1/x
    except:
        return Nothing()
    return Just(ret)

a = Just(2).bind(oneover) 
print(a)

a = Just(0).bind(oneover) 
print(a)
````

## Additional functions

There are some additional functions that are worth looking at too:

`accumulate` is similar to sum, but it keeps a running total of values.
    
`chain` joins two or more iterables into a single iterator.
    
    
`tee` returns n independent iterators from a single iterable.
    
    
`takewhile` is similar to filter; it returns all the values from an iterable until its function evaluates to false. Unlike filter, takewhile stops completely after the first false value.
    
`dropwhile` is the opposite of takewhile. It ignores all values until the function evaluates to false and then returns all values after that.
