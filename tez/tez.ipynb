```python
def get_df(file):
    prices_df = pd.read_csv(file,  parse_dates=['date'])
    #prices_df = prices_df.astype({'date': 'datetime64[ns, UTC]'}, copy=False)
    maxvol_df = prices_df[['date', 'vol']].groupby(prices_df['date'].map(lambda x: x.year)).agg('max')
    #maxvol_df['date'] = df_vol['date'].apply(lambda x: pd.Timestamp(x.year, x.month, x.day), axis=1)
    # I think index should be removed to pass auto tests...
    maxvol_df = maxvol_df.reset_index(drop=True)
    maxclose_df = prices_df[['date', 'close']].groupby(prices_df['date'].map(lambda x: x.year)).agg('max')
    maxclose_df = maxclose_df.reset_index(drop=True)
    return maxvol_df, maxclose_df

def solution(files):
    # files - any of available files, i.e:
    # files = ["./data/framp.csv", "./data/gnyned.csv", "./data/gwoomed.csv",
    #            "./data/hoilled.csv", "./data/plent.csv", "./data/throwsh.csv",
    #            "./data/twerche.csv", "./data/veeme.csv"]

    # write your solution here
    res = []
    for f in files:
        df_vol, df_close = get_df(f)
        res.append([df_vol, df_close])
    return res
```
